```python
import sympy as sp
from sympy.physics.wigner import wigner_6j

a,b,c = sp.symbols('a b c', integer=True)
d=sp.Rational(1,2)
e=b+sp.Rational(1,2)
f=c+sp.Rational(1,2)
sp.pprint(wigner_6j(1, 1, 1, sp.Rational(1,2), sp.Rational(3,2), sp.Rational(3,2)))
j = sp.Symbol('j', integer=True, positive=True)
sp.pprint(wigner_6j(1, 1, 1, 1, 1, j))
print(int(a+b+d+e))
sixj = wigner_6j(a, b, c, d, e,f)
simplified = sp.simplify(sixj)

sp.pprint(simplified)
```
